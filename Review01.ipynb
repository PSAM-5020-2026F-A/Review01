## Fancy calculator

Programming languages have built-in math capabilities.

Can easily do algebra ($+$, $-$, $\times$, $\div$, $2^2$, $\sqrt{2}$) with numbers and variables.

Sometimes we have to be careful about _floating point_ numbers versus whole numbers (_int_).

In `python` we can raise numbers to a power with `**`, so $4^3$ can be written as `4 ** 3`.

The square root of a number $\sqrt{1678}$ is the same as raising it to $\frac{1}{2}$, so `1678 ** 0.5`.

````python
# algebra
print((12 * 4 + 124 * 98) - 1234 * 5678 / 890)

# variables
x = 10 + 2
print(x)

y = 4 * 18283
print(y)

z = y - x
print(z)

# int/float/string

# this produces a float (number with a decimal point)
w0 = 10 / 3

# this truncates the result to an int (integer; whole number)
w1 = 10 // 3

# this is the text representation of the numeral "3"
w2 = "3"

# w1 will look the same as w2
print(w0, w1, w2)

# but they're different objects
print(w0+w0, w1+w1, w2+w2)
````

## Functions

Functions are pre-defined sequences of instructions for the computer to perform on a given set of inputs. The inputs can change, but the sequence of operations remain the same.

````python
# Function to add 2 numbers

# This is the function definition
#   given 2 numbers a and b, return the square of their difference
def diff_sq(a, b):
  return (b - a) ** 2

print(diff_sq(19, 87))

# Given a quadratic equation of the form:
#   A*x*x + B*x + C = 0
#   solve for x
def quadratic(A, B, C):
  sqrt_term = (B * B - 4 * A * C) ** 0.5
  sol0 = (-B + sqrt_term) / (2 * A)
  sol1 = (-B - sqrt_term) / (2 * A)
  return sol0, sol1

# solve: -2*x*x + 6x + 8
quadratic(-2, 6, 8)
````

## Using other people's code

We can easily import pre-built modules into our code

````python
# Run this cell to import libraries to load text and image from URLs

import requests  # for accessing URLs
import io        # for working with raw bytes
import PIL.Image # for working with images
````

Now we can make a request to get information about the Astronomy Picture of the Day, at this url:
`https://apod.as93.net/apod`.

````python
# We will revisit these again in class.
#   For now they're just examples of how to use these libraries

# Get data at the url
data_response = requests.get("https://apod.as93.net/apod")

# Read data as JSON
data = data_response.json()

# Display result
display(data)
display(data["url"])

# Get data from the url specified in the previous response
img_reponse = requests.get(data["url"])

# Read data as raw bytes
img_bytes = io.BytesIO(img_reponse.content)

# Transform bytes into image
img = PIL.Image.open(img_bytes)

# Display image
display(img)
````

````python
# Can use the same libraries to make image searches with code

import re # for processing text

query = "gudetama"

# Python f-string
#   to insert the content of a variable inside a string
query_url = f"https://www.bing.com/images/search?q={query}"

# get data, transform to text, display
data_response = requests.get(query_url)
data = data_response.text
display(data)

# This is a Regular Expression.
#   We'll see this in detail in the Data Structures class
#   For now, just a preview of using it to extract specific patterns from the text

# This looks for <img src=....>
pattern = r'<img[^>]+src\s*=\s*["\']([^"\']+)["\'][^>]*>'

# This looks for src=....
pattern = r"src\s*=\s*[\"']*(http\S+)[\"']"

# This performs the search and gives us a list of urls
#   embedded in the text of the search result
image_urls = re.findall(pattern, data, re.IGNORECASE)

# Loop through urls
for url in image_urls:
  # If url is not hosted in mm.bing.net,
  #   it's not an image url, so skip it
  if "mm.bing.net" not in url: continue

  # Get data, read as bytes, transform to image, display
  img_reponse = requests.get(url)
  img_bytes = io.BytesIO(img_reponse.content)
  img = PIL.Image.open(img_bytes).convert("RGB")
  display(img)
````
